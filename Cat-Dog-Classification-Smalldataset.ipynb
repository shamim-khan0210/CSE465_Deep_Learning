```python
import cv2
import os
import random
import numpy as np
```

```python
mydir = r'C:\Users\Md.Shamim Khan\OneDrive\Documents\CSE465_Deep Learning\DL\CNN\Datasets\train'
```

```python
categories = ['cats', 'dogs']
```

```python
data = []
for i in categories:
    folder_path = os.path.join(mydir,i)
    if i == 'cates':
        label = 0
    else:
        label = 1

    for j in os.listdir(folder_path):
        img_path = os.path.join(folder_path,j)
        img = cv2.imread(img_path)
        img = cv2.resize(img,(150,150))
        data.append([img,label])
```

```python
random.shuffle(data)
```

```python
X = []
y = []
for i in data:
    X.append(i[0])
    y.append(i[1])
```

```python
y = np.array(y)
```

```python
X = np.array(X)
```

```python
X.shape
```

```python
X = X/255
```

## Create CNN from here

```python
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Dense, Flatten, Activation, Dropout
```

```python
model = Sequential()
model.add(Input(shape=(150, 150, 3)))  # Define input shape explicitly

model.add(Conv2D(32, (3, 3)))
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Conv2D(32, (3, 3)))
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Conv2D(64, (3, 3)))
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))
```

```python
model.add(Flatten()) # This converts our 3D features map to 1D feature vectors
model.add(Dense(64))
model.add(Activation('relu'))
model.add(Dropout(0.5))
model.add(Dense(1))
model.add(Activation('sigmoid'))

model.compile(loss='binary_crossentropy',
             optimizer='rmsprop',
             metrics=['accuracy'])
```

```python
model.fit(X,y,epochs=5, validation_split=0.1)
```
