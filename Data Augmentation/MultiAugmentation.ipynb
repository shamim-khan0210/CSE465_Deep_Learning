```python
# Data Augmentation -> Create images from one image
from tensorflow.keras.preprocessing.image import ImageDataGenerator  
```

```python
batch_size = 16

# this is the augmentation configuration we will use for training 
train_datagen = ImageDataGenerator(
    rescale = 1./255,
    shear_range = 0.2,
    zoom_range = 0.2,
    horizontal_flip = True
)

# This is the augmentation configuration we will use for testing:
# Only rescaling
test_datagen = ImageDataGenerator(rescale=1./255)

# This is a generator that will read pictures found in
# subfolder of ''Datasets/train', and indefinitely generate
# batches of augmented image data
train_generator = train_datagen.flow_from_directory(
    'Datasets/train', # this is the target directory
    target_size = (150, 150), # all images resize to 150x150
    batch_size = batch_size,
    class_mode= 'binary'
) # since we use binary cross entropy loss, we need binary labels

# This is a similar generator, for validation data
validation_generator = test_datagen.flow_from_directory(
    'Datasets/validation',
    target_size=(150,150),
    batch_size= batch_size,
    class_mode='binary'
)
```

```python
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Dense, Flatten, Activation, Dropout

model = Sequential()
model.add(Input(shape=(150, 150, 3)))  # Define input shape explicitly

model.add(Conv2D(32, (3, 3)))
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Conv2D(32, (3, 3)))
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Conv2D(64, (3, 3)))
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))
```

```python
model.add(Flatten()) # This converts our 3D features map to 1D feature vectors
model.add(Dense(64))
model.add(Activation('relu'))
model.add(Dropout(0.5))
model.add(Dense(1))
model.add(Activation('sigmoid'))

model.compile(loss='binary_crossentropy',
             optimizer='rmsprop',
             metrics=['accuracy'])
```

```python
model.fit(
    train_generator,
    steps_per_epoch=1888 // batch_size,
    epochs=25,
    validation_data=validation_generator,
    validation_steps=500 // batch_size
)
```
